```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
x = [1,2,3,4,5,10,15]
y = [6,7,2,9,10,20,6]

c = 'r'
plt.plot(x,y, c)
plt.title('Line graph using Matplotlib')
plt.show()
```

```python
 varr1 = [1,2,3,4,5,10,15]
varr2= [6,7,2,9,10,20,6]

sns.lineplot( x =  varr1, y =  varr2, color = 'red')
plt.title('Seaborn')
plt.show()
```

```python
varr1 = [1,2,3,4,5,10,15]
varr2= [6,7,2,9,10,20,6]

fig = px.line( x = varr1, y = varr2 ,title = 'Plolty') #, marker = '*')
fig.show()
```

```python
import plotly.graph_objects as go 
```

```python
fig = go.Figure(go.Scatter(x = varr1, y = varr2, ))
fig.show()
```

```python
fig.update_layout(title = 'Plotly graph' , xaxis_title = 'This is x axis', yaxis_title  = 'This is Y axis')
fig.show()
```

```python
varr1x,varr1y = [1,2,3,4],[3,5,6,7]
varr2x,varr2y = [6,7,2,4],[5,6,7,8]
varr3x,varr3y = [7,8,9,6],[11,12,13,4]

fig = go.Figure()

fig.add_trace(go.Scatter(x = varr1x, y = varr1y))
fig.add_trace(go.Scatter(x = varr2x, y = varr2y))
fig.add_trace(go.Scatter(x = varr3x, y = varr3y))
fig.show()
```

```python
import plotly.express as px
import pandas as pd

# Example dataframe (तुम्हारे पास पहले से data होगा)
data = pd.DataFrame({
    "Product Name": ["iPhone 12", "iPhone 13", "iPhone 14", "iPhone 15"],
    "Number Of Ratings": [1200, 1500, 1800, 2000]
})

# Bar graph बनाना
fig = px.bar(
    data,
    x="Product Name",          # X-axis पर product names
    y="Number Of Ratings",     # Y-axis पर ratings count
    title="iPhone Ratings Count", 
    width=800,
    height=500,
    color="Product Name"       # अलग-अलग रंग products के लिए
)

fig.show()
```
