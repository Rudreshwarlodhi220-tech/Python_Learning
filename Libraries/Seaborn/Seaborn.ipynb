# Seaborn - 
- Seaborn is a library for making statical graph in python .
- Seaborn builds on Top of Matplotlib and integrated closely with Panda data structure.
  (सीबॉर्न, Matplotlib के ऊपर बनाया गया है और ।)
- Seaborn helps you explore and understand your data. Its floating function operation on dataFrames and arrays containing whole data sets internally perform new Symantec mapping and statistical aggregation to produce improved plots .
- Plots are basically used for visualizing the relations between variables those variables be either completely numerical or a category like a group, class, or division. Seborn divides the plot into the blow categories.
- 
  1. Relationship plots : This plot is used to understand the relation between two variables.
  2. Categorical plots : This plot deals with categorical variables and how they can be visualize.
  3. Distribution plots : This plot is used for examining univariate and bivariety distribution.

## Bar plot in Seaborn

```python
import seaborn as sns 
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset('penguins')
df
```

```python
df[['island']]
```

```python
g = sns.barplot(x = df.island , y = df.bill_length_mm)
plt.show()
```

Line plots (): These are typically used to show trends or relationships between two continuous variables.
    In your example,  vs.  makes sense because both are numeric measurements, and you’re exploring how they vary 
    together.
• 	Bar plots (): These are usually used to compare categories against a numeric variable. 
    That’s why you set  (categorical) and  (numeric). The bar plot then shows average bill length for each island.
👉 So the difference isn’t about the plotting function itself, but about the data types:


```python
g=sns.barplot(x = df.island , y = df.bill_length_mm ,data = df)
```

```python
sns.barplot(x = 'island', y = 'bill_length_mm' , data = df)
plt.show()
```

```python
sns.barplot(x = 'island', y = 'bill_length_mm' , data = df , hue = 'sex' , )
plt.show()
```

## Order

```python
order1 = ['Dream','Torgersen','Biscoe']  # order change ourself .

sns.barplot(x = 'island', y = 'bill_length_mm' , data = df, hue = 'sex' , order = order1 )
plt.show()
```

## hue_order

```python
order1 = ['Dream','Torgersen','Biscoe']  # convert Femal = blu and ... .

sns.barplot(x = 'island', y = 'bill_length_mm' , data = df, hue = 'sex' ,
            order = order1 ,hue_order = ['Female','Male'])
plt.show()
```

```python
order1 = ['Dream','Torgersen','Biscoe']
sns.barplot(x ='island' , y = 'bill_length_mm', data = df , color = 'green')
plt.show()
```

```python
order1 = ['Dream','Torgersen','Biscoe']
sns.barplot(x = 'island' ,y = 'bill_length_mm', data = df , hue = 'sex',
           palette = 'GnBu')
```

## Saturation 

```python
order1 = ['Dream','Torgersen','Biscoe']
sns.barplot(x = 'island' ,y = 'bill_length_mm', data = df , hue = 'sex',
           palette = 'GnBu', saturation = 0.5)   # use to get dull.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [12,15,18,22,25,28,30,32,35,38,40,42,45,48,50,52,55,58,60,62] # total number = 20 means 5 no ka 1bin bana sakte hai(block). 

# Histogram with 5 bins
sns.histplot(marks, bins=5) # yaha 5 block me devide krr diya hai [12,15,18,22,25]..[].[]..[]..[].
plt.show()

# Histogram with custom bins
sns.histplot(marks, bins=[10,20,30,40,50,60,70])
plt.show()
```
