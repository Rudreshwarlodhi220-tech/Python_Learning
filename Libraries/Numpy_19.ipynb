# Sorting

```python
first_arr = np.array([3,6,7,6,8,8,])
print(first_arr,'\n')

sort_first_arr = np.sort(first_arr)
print(sort_first_arr)
```

```python
two_arr = np.array([[3,523,65],[46,463,6]])
print(two_arr)
print("_"*50)
sort_two_arr = np.sort(two_arr)
print(sort_two_arr)
print("_"*50)

sort_two_arr = np.sort(two_arr,axis=0)
print(sort_two_arr)
print("_"*50)

sort_two_arr = np.sort(two_arr,axis=1)
print(sort_two_arr)
```

axis=0 → इसका मतलब है rows के ऊपर काम करना, यानी column-wise operation.

जब तुम np.sort(..., axis=0) लिखते हो, तो NumPy हर column को अलग-अलग sort करता है।

Rows की positions बदलती हैं ताकि हर column sorted हो जाए।

axis=1 → इसका मतलब है columns के ऊपर काम करना, यानी row-wise operation.

जब तुम np.sort(..., axis=1) लिखते हो, तो NumPy हर row को अलग-अलग sort करता है।

Columns की positions बदलती हैं ताकि हर row sorted हो जाए।
