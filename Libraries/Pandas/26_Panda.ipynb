## hamesa loc.[] ya iloc.[] hi use karna hai kisi bhi operation ko pperform karne ke liye kyuki panda kab view return kare or kab actual value change kre de ye nahi pata rahta  hai .

```python
new_df.loc[[1,2],['C','D']]    # When we want to only specific row and column.

# loc = location based indexing .
```

```python
new_df.loc[:,['C','D']]  # when we want all C,D rows.
```

```python
new_df.loc[[1,2] , :]  # When we want to all columns OF 1 & 2.
```

```python
new_df.loc[(new_df['A'] < 0.3 )] # 
```

```python
new_df.loc[(new_df['A'] < 0.3 ) & (new_df['C'] > 0.2) ] # It doesn't change actual value.
```
