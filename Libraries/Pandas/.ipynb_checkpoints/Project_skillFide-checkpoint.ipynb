```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
# Identify the Target Variable
# In this dataset, the target column is usually y, which indicates whether the client subscribed to a term deposit (yes or no).
# Check.
df['y'].value_counts()

```

```python
# Plot Distribution of Target Variable
# If Target is Categorical (yes/no)


sns.countplot(x=df['y']) # x=df['y'] it explain in which column distribution of the target variable
plt.title("Distribution of Target Variable (Term Deposit Subscription)")
plt.xlabel("Subscription (y)")
plt.ylabel("Count")
plt.show()

```

```python
# If Target is Numeric .
sns.histplot(df['y'], kde=True)
plt.title("Distribution of Target Variable")
plt.xlabel("Target")
plt.ylabel("Frequency")
plt.show()
```

```python
Step 4: Quick Insight
- In the Bank Marketing dataset, the target variable y is imbalanced — usually many more no than yes.
- That imbalance is important to notice because it affects model training (you may need techniques like oversampling, undersampling, or class weights).

✅ So, the solution is:
- Use countplot for categorical targets (yes/no).
- Use histplot for numeric targets.

Would you like me to also show you how to combine the plot with percentages (so you see not just counts but also the proportion of yes vs no)? That’s often very useful when dealing with imbalanced datasets like this one.
```
