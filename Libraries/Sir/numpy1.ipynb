# Zero Dimensional Array

```python
zero_arr = np.array(10)
print(zero_arr)
print(type(zero_arr))
print(f"Dimension of an array {zero_arr.ndim}")
```

# One Dimension Array

```python
one_d_arr = np.array([10])
print(one_d_arr)
print(type(one_d_arr))
print(f"Dimension of an array {one_d_arr.ndim}")
```

```python
one_d_arr = np.array([10,6,3,6,7,])
print(one_d_arr)
print(type(one_d_arr))
print(f"Dimension of an array {one_d_arr.ndim}")
```

```python
two_d_arr = np.array([[10,6],[5,7]])
print(two_d_arr)
print(type(two_d_arr))
print(f"Dimension of an array {two_d_arr.ndim}")
print(f"Shape of an array {two_d_arr.shape}")
print(f"Data Type of an array {two_d_arr.dtype}")
```

```python
two_d_arr = np.array([[10.56,6],[5,7]])
print(two_d_arr)
print(type(two_d_arr))
print(f"Dimension of an array {two_d_arr.ndim}")
print(f"Shape of an array {two_d_arr.shape}")
print(f"Data Type of an array '{two_d_arr.dtype}'")
```
