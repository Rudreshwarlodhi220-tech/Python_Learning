## --- request - 
    - Python has an HTTP library that is used to interact with web pages..

    - It sends a request to the website (like when you open a page from a browser).

    - Website ek response bhejti hai (HTML, JSON, text, etc.), jise hum parse karke data nikal sakte hain.

    

    - The website sends a response (HTML, JSON, text, etc.), which we can parse to extract data.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops"
```

```python
response = requests.get(url)
print(response)                # [100] , [200] means no erro if 400 in output means error.
```

```python
response.content # When we want to see the whole code .
```

```python
# soup is a variable.
soup = BeautifulSoup(response.text,'html.parser') # parser maens  
```

- Parser ek tool hai jo raw HTML ko tod‑fod kar ek structured format (tree structure) mein convert karta hai.

- Website se jo HTML aata hai wo ek plain text string hota hai. Usko directly samajhna mushkil hai.

- Parser us text ko readable structure mein badal deta hai jisme tum easily tags (<title>, <p>, <a>) search aur extract kar sakte ho.

```python
print(soup)
```

## -- When we want pro name , price and ratiing from the website .

## --- Extract Name

```python
# Here we use find_all method of(ka) Beautifulsoup .
Pro_name = soup.find_all('a',class_ = 'title')
Pro_name_lst = []

for pro_name in Pro_name:
    title = pro_name.text     # bass text nikalane ke liye use kiya hai .
    Pro_name_lst.append(title)
print(Pro_name_lst)
```

```python
0
```

```python
Pro_name = soup.find_all('a',class_ = 'title')
Pro_name_lst = []

for pro_name in Pro_name:
    title = pro_name.text
    Pro_name_lst.append(title.strip()) # strip use to remove /t,/n,/b
print(Pro_name_lst)
```

## --- Extract Discription

```python
Pro_discription = soup.find_all('p',class_ = 'description card-text')
Pro_discription_lst  = []

for pro_discrip in Pro_discription:
    discription = pro_discrip.text
    Pro_discription_lst.append(discription.strip())

print(Pro_discription_lst)
```

## ---- Extract Price 

```python
Pro_price = soup.find_all('h4',class_ = 'price float-end card-title pull-right')
Pro_price_lst = []

for pro_price in Pro_price:

    price = pro_price.text
    Pro_price_lst.append(price.strip())

print(Pro_price_lst)
```

## Review-Rating

```python
Review_rat = soup.find_all('p',class_ = 'review-count float-end')
Review_rating_lst = []

for re_rat in Review_rat:
    rating = re_rat.text
    Review_rating_lst.append(rating.strip())

print(Review_rating_lst)
```

```python
import pandas as pd
df  = pd.DataFrame({ 'Product_name':Pro_name_lst ,
                    'Product_discription':Pro_discription_lst,
                    'Product_price':Pro_price_lst,
                    'Review_Rating':Review_rating_lst
})

print(df)
```
