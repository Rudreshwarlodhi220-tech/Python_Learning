# --- Fill Null values using for loop 

```python
import pandas as pd
pengu_df = pd.read_csv(r'D:\AI & ML\Libraries\Sir\Pandas\Dataset\penguins.csv')
pengu_df
```

## Merge - 
        - Merge me do table me ak same column hona chahiye.

```python
import pandas as pd
df_stu = pd.DataFrame({
                        'Stu_id':[101,102,104,105],
                        'Name':['Rakesh','Manav','Harshita','Astha'],
})

df_academics = pd.DataFrame({ 'Stu_id':[101,102,103,104],
                             'Percentage':[89,44,79,23]
                             })

```

```python
merg_df = df_stu.merge(df_academics , on='Stu_id')
merg_df
```

```python
merg_df = df_stu.merge(df_academics , on='Stu_id', how = 'inner')  # inner na bhi de to ye by default lega.
merg_df
```

```python
df_merge = df_stu.merge(df_academics , on = 'Stu_id' , how = 'outer')
df_merge
```
