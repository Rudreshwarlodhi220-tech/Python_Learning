```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

# Line Chart 

```python
arr1 = [3,6,3,6,7,4,7,7,]
plt.figure(figsize=(5,2))
plt.plot(arr1)
plt.title("Line Chart")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
arr1 = [3,6,3,6,7,4,7,7,]
plt.figure(figsize = (5,2))
sns.lineplot(arr1)
plt.show()
```

```python
arr1 = [3,6,3,6,7,4,7,7,4]
arr2 = [5,3,5,7,8,6,8,9,9]
arr3 = [3,2,5,5,6,5,7,8,6]
plt.figure(figsize=(5,2))
plt.plot(arr1,label = "arr1",c = 'red',ls = "--", marker="o")
plt.plot(arr2,label = "arr2",c = 'black',ls = "dotted")
plt.plot(arr3,label = "arr3",c = 'blue',ls = "-.")
plt.title("Line Chart")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.legend()
plt.show()
```

```python
arr1 = [3,6,3,6,7,4,7,7,4]
arr2 = [5,3,5,7,8,6,8,9,9]
plt.figure(figsize=(5,2))
sns.lineplot(arr1,label = 'arr1')
sns.lineplot(arr2,label = 'arr2')
plt.legend()
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(5,3))
plt.bar(x = cate,height=value)
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(5,3))
plt.barh(y = cate,width=value)
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(5,3))
sns.barplot(x = cate,y=value,hue = cate)
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(6,3))
sns.barplot(x = cate,y=value,hue = cate,orient='h')
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(3,3))
plt.pie(x = value,labels=cate,autopct="%0.2f%%",)
plt.legend()
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(3,3))
plt.pie(x = value,labels=cate,autopct="%0.2f%%",
        startangle=90,shadow=True,explode = [0.1,0.1,0.1])
plt.legend()
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(3,3))
plt.pie(x = value,labels=cate,autopct="%0.2f%%",
        startangle=90,shadow=True,explode = [0.1,0.1,0.1],
        counterclock = False,wedgeprops=dict(width = 0.5))
plt.legend()
plt.show()
```

```python
cate = ['A',"B",'C']
value = [450,300,500]
plt.figure(figsize=(3,3))
plt.pie(x = value,labels=cate,autopct="%0.2f%%",
        wedgeprops=dict(width = 0.6))
plt.legend()
plt.show()
```
