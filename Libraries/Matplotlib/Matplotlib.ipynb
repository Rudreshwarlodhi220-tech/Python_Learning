# Matplotlib ek Python library hai jo mainly data visualization ke liye use hoti hai. Matlab tumhare paas agar data hai (CSV, SQL se fetched, Pandas DataFrame, etc.), to Matplotlib us data ko graphs aur charts ke form me dikhane ke liye powerful tool hai.

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [6,7,2,9,10]


plt.plot(x,y)
# plt.show()
```

```python
x = [1,2,3,4,5]
y = [6,7,2,9,10]

c = 'green'

plt.plot(x,y,c)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [6,7,2,9,10]


plt.plot(x,y , marker = '*')
plt.show()
```

```python
x = [1,2,9,4,3]
y = [6,7,2,9,10]

plt.plot(x,y)

plt.plot(x, marker = '*')

plt.plot(y, marker = '^')
plt.show()
```

# Bar graph

```python
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([6,7,2,9,10])

colorr = 'red','blue','green','yellow','purple'

plt.bar(x,y , color = colorr, width= 0.8)

plt.show()
```

```python
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([6,7,2,9,10])

colorr = 'red','blue','green','yellow','purple'
plt.barh(x,y , color = colorr, height= 0.2)

plt.show()
```

# Scatter plot

```python
x = [1,2,3,4,5]
y = [6,7,2,9,10]

colorr = 'red','blue','green','yellow','purple'
plt.scatter(x,y , color = colorr, s = 250)

plt.title('Scatter plot')
plt.xlabel('X axis')
plt.ylabel('y axis')

plt.show()
```

# Pie chart
