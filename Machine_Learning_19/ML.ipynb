# ML lifecycle - 
- ### a) Data gathering/Collection -
        
        -        Own Dataset (स्वयं का डेटा)

                खुद collect किया हुआ या manually बनाया हुआ dataset।

                Example: Survey forms, experiment results, या manually curated data.

        -      Client-Based Dataset (क्लाइंट का डेटा)

                Client या organization द्वारा provide किया गया dataset।

                Example: किसी company का sales data, customer records.

        -        Database (डेटाबेस से डेटा)

                Existing databases से data निकालना।

                Example: SQL, PostgreSQL, MongoDB से stored data fetch करना।

        -        Online Dataset (ऑनलाइन उपलब्ध डेटा)

                Publicly available datasets जो internet पर free या paid मिलते हैं।

                Example: Kaggle, UCI Machine Learning Repository, Government portals.
