# Decorators -
- Decorators are powerful tools that allow us to modify or extend the behavior of function without changing their actual code They are the that warp another function adding same kind of functionality before or after the warped function is called.



```python
def greet(func):
    def wrapper():    # wraper ki place prr or kuch bhi rakh sakte hai
        print('Welcome kode gurukul..')
        print('Thanks..')
        func()
    return wrapper # return throw karega @greet catch karega. tab wrapper convert into helo_msg.

@greet
def helo_msg():
    print('Hello how are  you..')

helo_msg()
```

```python
def greet(fun):
    def wraper():
        print('Welcome Harsh\n\n')
        fun()
        print('thanks\n\n')
    return wraper

def msg_g():
    print('I will come at 5 PM')

def eat_msg():
    print('Food store in fridge')


@greet
def pet_msg():
    print('You have to go for walk with toomy')
    msg_g()
    eat_msg()

pet_msg()
```

```python
def decorator(func):
    def wrapper():
        print('Befor function call')
        func()
        print('After function call')
    return wrapper

def hello_msg():
    print("Hello everyone")

hello_msg = decorator(hello_msg)
hello_msg()
```

```python
def decorater(fun):
    def wrapper(a,b):
        print("Addition of 2 numbers = ",end='')
        fun(5,9)
    return wrapper

@decorater
def addition(a,b):
    print(a+b)

addition(5,9)
```

```python
def log_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with {args} {kwargs}")
        return func(*args, **kwargs)
    return wrapper

@log_decorator
def add(a, b):
    return a + b

print(add(3, 4))
```

```python
def require_login(func):
    def wrapper(user):
        if not user.get("logged_in"):
            print("Access Denied!")
            return
        return func(user)
    return wrapper

@require_login
def dashboard(user):
    print("Welcome to your dashboard!")

dashboard({"logged_in": False})
```
