# Time Module -
 - The time module in Python provide various time related function. It allow us to view the time functionalities like getting the current time pushing the program for a certain duration and formatting time.

```python
import time
```

```python
print(time.ctime())
```

```python
print(time.localtime())  # show all M,Y,D,T,S...

print(time.localtime().tm_mon)
print(time.localtime().tm_min)
```

```python
print(1)
time.sleep(5)
print(2)
```

## time.time() return to current time in second since the epoch (January 1 1970).

```python
current_time = time.time()

print(current_time)
```

```python
t = time.time()

print(1)
time.sleep(5)
print(2)

time.time()-t
```

```python
def claculate_time():
    print(1)
    time.sleep(5)
    print(2)
t = time.time()

claculate_time()
print(time.time()-t)
```

```python
def time_calculation():
    print(1)
    time.sleep(5)
    print(2)
t = time.time()

time_calculation()

print(time.time()-t)
```
