#   ------------ Generators -----------
- Generator in Python is a special type of iterable like list or tuple but they do not store all their values in our memory They generate values on the fly. They are the function that returned an irritator that we can itreate over( one value at a time). They are written using the yield keyword which make the function a generator.

```python
def squar():
    lst = []
    for i in range(1,100):               # with out generator.
        lst.append(i**2)
    return lst                       # yaha ye aak bbaar me sb ko hi print krr de raha hai
#     print(lst)

# squar()

b = squar()
print(b)
```

```python
def sqr():
    list = []
    for i in range(1,10):
        list.append(i**2)
    return list


sqr() 
```

```python
def num():
    for i in range(0,100):
        yield i

b = num()


print(next(b))
print(next(b))
```

```python
def squar():
    a = list(range(1,100))

    for i in a :
        yield i**2    # agar koii file jisme bahut sara data hai or hum ussko line by line load kare

b = squar()
print(next(b))   # iska mtlb ye aak time  prr ak hi generate karna hai.
print(next(b))
print(next(b))
print(next(b))


# for i in range(1,20):
#     print(next(b))
```

```python
def infinite():
    count = 0
    while True:       # infinite loop bhi system ko harm nahi krr sakta...
        yield count

        count+=1
f =  infinite()

print(next(f))
print(next(f))
print(next(f))
print(next(f))
print(next(f))
print(next(f))

# for i in range(1,10):
#     print(next(f))
```

```python
def infinite_numbers():
    n = 1
    while True:   # Infinite loop
        yield n
        n += 1

# Generator से पहले 10 numbers लेना
gen = infinite_numbers()
for _ in range(10):
    print(next(gen))
```

```python
def infinite_even():
    n = 0
    while True:
        yield n
        n += 2

gen = infinite_even()
for _ in range(8):
    print(next(gen))
```
