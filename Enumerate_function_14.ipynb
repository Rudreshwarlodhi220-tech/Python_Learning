# Enumerate Function -
- The enumerate function in Python adds counter to an iter and return it as an enumerate object This object can then be used directly in for loop or converted into a list of tuple.

```python
my_lst = ['a','b','c','d','e','f']

for ind,i in enumerate(my_lst):
    print(ind,i)
```

```python
my_lst = ['a','b','c','d','e','f']

for index,i in enumerate(my_lst,100): # 100 usko index di hai ki chalana kaha se hai.
    print(index,i)
```

```python
my_lst = ['a','b','c','d','e','f']

print(tuple(enumerate(my_lst,1)))
```
