# Indexing and Slicing 

```python
one_arr = np.array([3,6,7,6,8,8,])
print(one_arr)
print(one_arr[0])
print(one_arr[1])
print(one_arr[4])
print(one_arr[5])
print(one_arr[-3])

print(one_arr[2:5])
print(one_arr[3:5])
print(one_arr[4:1:-1])
print(one_arr[::-1])
```

```python
two_arr = np.array([[3,6,7],[6,8,8,]])
print(two_arr)
print(two_arr[0])
print(two_arr[0][0])
print(two_arr[0][1])
print(two_arr[0][2])
print(two_arr[1][0])
print(two_arr[1][1])
print(two_arr[1][2])

print(two_arr[1][:2])
print(two_arr[1][1:])
print(two_arr[1][::-1])
```

```python
three_arr = np.array([[[3,6,7],[6,8,8,],[45,23,2]]])
print(three_arr)
print(three_arr[0])
print(three_arr[0][1])
print(three_arr[0][1][0])
```

```python
three_arr = np.array([[[3,6,7],[6,8,8,]],[[45,23,2],[89,34,45]]])
print(three_arr)
print(three_arr[0])
print(three_arr[0][1])
print(three_arr[0][1][0])
```

# Mathematical and Statistical Operation

```python
first_arr = np.array([3,6,7,6,8,8,])
print(first_arr)
second_arr = np.array([2,4,6,6,7,12])
print(second_arr)
```

```python
Add = first_arr + second_arr
print(Add)
```

```python
Addition = np.add(first_arr,second_arr)
print(Addition)
```

```python
Sub = first_arr - second_arr
print(Sub)
```

```python
Substraction = np.subtract(first_arr,second_arr)
print(Substraction)
```

```python
Multi = first_arr * second_arr
print(Multi)
```

```python
Multiplication = np.multiply(first_arr,second_arr)
print(Multiplication)
```

```python
div = first_arr / second_arr
print(div)
```

```python
division = np.divide(first_arr,second_arr)
print(division)
```

```python
new_a = np.array([[3,6],[6,8]])
new_b = np.array([[13,3],[7,3]])
print(new_a)
print("--"*50)
print(new_b)
mat_multi = np.matmul(new_a,new_b)
print(mat_multi)
```

```python
new_a = np.array([[3,6],[6,8]])
new_b = np.array([[4,3],[7,3]])
print(new_a)
print("--"*50)
print(new_b)
mat_multi = np.matmul(new_a,new_b)
print(mat_multi)
```

```python
first_arr = np.array([3,6,7,6,8,8,])
print(first_arr)
print(first_arr.min())
print(first_arr.max())
print(first_arr.mean())
print(first_arr.var())
print(first_arr.std())
print(first_arr.argmin())
print(first_arr.argmax())
```

```python
first_arr = np.array([3,6,7,6,8,8,])
min = np.min(first_arr)
max = np.max(first_arr)
mean = np.mean(first_arr)
median = np.median(first_arr)
var = np.var(first_arr)
std = np.std(first_arr)

print(min,max,mean,median,var,std,)
```

# Sorting

```python
first_arr = np.array([3,6,7,6,8,8,])
print(first_arr)
sort_first_arr = np.sort(first_arr)
print(sort_first_arr)
```

```python
two_arr = np.array([[3,523,65],[46,463,6]])
print(two_arr)
print("_"*50)
sort_two_arr = np.sort(two_arr)
print(sort_two_arr)
print("_"*50)
sort_two_arr = np.sort(two_arr,axis=0)
print(sort_two_arr)
print("_"*50)
sort_two_arr = np.sort(two_arr,axis=1)
print(sort_two_arr)
```
