```python
class InvalidAgeerror(Exception):    # yaha InvaliA.. ak custom exception hai .then sabse pahle Exception kko inherit karenge .
    pass
age = int(input('Enter your age = '))
try:
    if age<3:
        raise InvalidAgeerror      # ye kya kr raha hai ki jab bhi condition true hogi ye use
    else:                          # except block me send kar dega .
        print('You are elligible.')

except InvalidAgeerror:           # InvalidAgeerror ke threw catch kareega .
    print('You are  not eligible .' )
```

```python
class AgeLessError(Exception):
    def __init__(self,age):
        self.age = age

    def error_msg(self):
        print("Your age is Less than condition -----")

class AgeMoreError(Exception):   # We can use many classs ..
    def __init__(self,age):
        self.age = age

    def error_msg(self):
        print("Your age is more than condition >>>>")


age = int(input('Enter your age = '))
try:
    if age<3:
        raise AgeLessError(age)

    elif age>60:
        raise AgeMoreError(age)

    else:
        print('You are elligible.')

except AgeLessError as le:
        le.error_msg()
except AgeMoreError as Ae:
        Ae.error_msg()
```

```python
class Agelesserror(Exception):
    def __init__(self,age):
        self.age = age

    def error_msg(self):
        print("your age is less then condition")

class Agemoreerro(Exception):
    def __init__(self,age):
        self.age = age

    def error_msg(self):
        print("Your age is greter then condition")

age = int(input('Enter your age = '))
try:
    if age<3:
        raise Agelesserror(age)
    elif age>60:
        raise Agemoreerro(age)

except Agelesserror as le:
    le.error_msg()

except Agemoreerro as mo:
    mo.error_msg()
```
